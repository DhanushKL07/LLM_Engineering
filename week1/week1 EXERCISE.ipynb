```python
system_prompt = """
You are a helpful assistant that analyzes the question asked ,
and solve the question. Respond in less number of words.
Explain line by line
Respond in markdown. Do not wrap the markdown in a code block - respond just with the markdown.
"""
```

```python
# constants
openai=OpenAI
MODEL_GPT = 'openai/gpt-oss-120b'
MODEL_LLAMA = 'llama3.1:8b'
```

```python
# set up environment
openai = OpenAI(
    api_key=os.getenv("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",
)
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why,
response = openai.chat.completions.create(
    model="MODEL_GPT",
    messages=[
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": question}
"""
```

```python
# Get openai/gpt-oss-120b to answer, with streaming
stream = openai.chat.completions.create(
        model='openai/gpt-oss-120b',
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
          ],
        stream=True
    )    
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# Get llama3.1:8b to answer
openai = OpenAI(
    api_key='ollama',
    base_url="http://localhost:11434/v1",
)
stream = openai.chat.completions.create(
        model='llama3.1:8b',
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
          ],
        stream=True
    )    
response = ""
display_handle = display(Markdown(""), display_id=True)
for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        update_display(Markdown(response), display_id=display_handle.display_id)
```
